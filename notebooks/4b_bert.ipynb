# Environment

```python
import os
os.chdir("..")
```

```python
import pandas as pd
from src.pipeline.bert import BertMultilabelConfig, PipelineBertMultilabel

from src.utils.map import SYMPTOM_MAP
```

```python
MODEL_PATH = "./models/BERTimbau_depressive_symptoms"
```

## Data Load

```python
df = pd.read_csv("./data/processed/amive.csv")
```

# Model

## Config

```python
config = BertMultilabelConfig(
    model_path=MODEL_PATH,
    label_columns=list(SYMPTOM_MAP.values()),
    train_args={
        "num_train_epochs": 10,
        "per_device_train_batch_size": 16,
        "learning_rate": 2e-5,
        "warmup_ratio": 0.1,
        "weight_decay": 0.01,
        "eval_strategy": "epoch",
        "save_strategy": "epoch",
        "load_best_model_at_end": True,
        "metric_for_best_model": "eval_macro_avg_f1",
        "logging_strategy": "epoch",
        "report_to": "none",
    }
)
```

## Pipeline

```python
pipeline = PipelineBertMultilabel(config)
```

### Data Split 

| Set | % |
| :--- | :---: |
| **Train** | 65 |
| **Val** | 20 |
| **Test** | 15 |

Divisão estratificada, que preserva a distribuição de cada label e suas combinações entre os splits

```python
train_df, val_df, test_df = pipeline.split(df)
```

```python
trainer = pipeline.train(train_df, val_df, text_column="TEXT")
```

### Results

```python
results = pipeline.evaluate(trainer, test_df, text_column="TEXT")
```

```python
display(results)
```

### Save Model

```python
pipeline.save(trainer)
```

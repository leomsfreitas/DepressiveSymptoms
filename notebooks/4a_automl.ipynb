# Environment

```python
import os
os.chdir("..")
```

```python
import pandas as pd
from pathlib import Path
from src.pipeline.automl import AutoGluonConfig, PipelineAutoGluon

from src.utils.map import SYMPTOM_MAP
```

```python
MODEL_PATH = "./models/AutoGluon_depressive_symptoms"
```

## Data Load

```python
df_base = pd.read_csv("./data/processed/amive.csv")
df_lex = pd.read_csv("./data/features/amive_lex.csv")
df_ge = pd.read_csv("./data/features/amive_ge.csv")
df_gep = pd.read_csv("./data/features/amive_gep.csv")
```

# Model

```python
config = AutoGluonConfig(
    model_path=Path(MODEL_PATH),
    label_columns=list(SYMPTOM_MAP.values()),
    presets="medium_quality",
    time_limit=900,
)
```

## Train without Lex (Baseline)

### Config

```python
config.model_path = Path(MODEL_PATH, "baseline")
```

```python
baseline = PipelineAutoGluon(config, df_base)
```

### Data Split 

| Set | % |
| :--- | :---: |
| **Train** | 65 |
| **Val** | 20 |
| **Test** | 15 |

```python
train_df_base, val_df_base, test_df_base = baseline.split(df_base)
```

### Train

```python
baseline.train(train_df_base, val_df_base)
```

### Results

```python
results_base = baseline.evaluate(test_df_base)
```

```python
display(results_base)
```

### Save Model

```python
baseline.save()
```
